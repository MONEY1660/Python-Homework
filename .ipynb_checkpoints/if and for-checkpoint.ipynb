```python

score = int(input("Enter your score : "))
if score >= 80:
    print("Grade A")
elif score >= 70:
    print("Grade B")
elif score >= 60:
    print("Grade C")
elif score >= 50:
    print("Grade D")
else:
    print("Grade E")
    
try :
    score = int(input("enter score : "))
    
except :
    print("error")
```

```python
for i in range(0, 4):
    for j in range(0, 5):
        if i == 0 or i == 3:
            print("*", end="")
        elif j == 0 or j == 4:
            print("*", end="")
        else:
            print(" ", end="")
    print()


```
